```python

# Count employees by department and attrition

department_attrition = pd.crosstab(
    df_clean['Department'],
    df_clean['Attrition']
)

print(department_attrition)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import glob

# Find CSV file
files = glob.glob(
    '/content/employee_data/**/*.csv',
    recursive=True
)

print("CSV files found:", files)

# Load dataset
df = pd.read_csv(files[0])

# Create cleaned dataset
df_clean = df.drop(
    columns=['EmployeeCount', 'Over18', 'StandardHours'],
    errors='ignore'
).copy()

# Create screenshot folder
os.makedirs('/content/Screenshots', exist_ok=True)

print("Dataset loaded successfully!")
print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)
```

```python
salary_analysis = df_clean.groupby(
    'Attrition'
)['MonthlyIncome'].mean()

print(salary_analysis)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=salary_analysis.index,
    y=salary_analysis.values
)

plt.title('Average Monthly Income by Attrition Status')
plt.xlabel('Attrition Status')
plt.ylabel('Average Monthly Income')

plt.savefig(
    '/content/Screenshots/salary_attrition.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python

import os
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Create screenshot folder
os.makedirs('/content/Screenshots', exist_ok=True)

# GRAPH 1: Overall Attrition Distribution
plt.figure(figsize=(8, 5))
sns.countplot(data=df_clean, x='Attrition')
plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition Status')
plt.ylabel('Number of Employees')
plt.savefig('/content/Screenshots/attrition_distribution.png',
            dpi=300, bbox_inches='tight')
plt.show()


# GRAPH 2: Department-wise Attrition Rate
dept = pd.crosstab(df_clean['Department'], df_clean['Attrition'])

dept['Total'] = dept['Yes'] + dept['No']
dept['Attrition Rate (%)'] = dept['Yes'] / dept['Total'] * 100

plt.figure(figsize=(9, 5))
sns.barplot(x=dept.index, y=dept['Attrition Rate (%)'])
plt.title('Department-wise Employee Attrition Rate')
plt.xlabel('Department')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=15)
plt.savefig('/content/Screenshots/department_attrition.png',
            dpi=300, bbox_inches='tight')
plt.show()


# GRAPH 3: Salary vs Attrition
salary = df_clean.groupby('Attrition')['MonthlyIncome'].mean()

plt.figure(figsize=(8, 5))
sns.barplot(x=salary.index, y=salary.values)
plt.title('Average Monthly Income by Attrition Status')
plt.xlabel('Attrition Status')
plt.ylabel('Average Monthly Income')
plt.savefig('/content/Screenshots/salary_attrition.png',
            dpi=300, bbox_inches='tight')
plt.show()


print("All three graphs saved successfully!")
print(os.listdir('/content/Screenshots'))
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

folder = '/content/Screenshots'
os.makedirs(folder, exist_ok=True)

# 1. Work-Life Balance
balance = pd.crosstab(
    df_clean['WorkLifeBalance'],
    df_clean['Attrition']
)

balance_rate = balance['Yes'] / balance.sum(axis=1) * 100

plt.figure(figsize=(8, 5))
sns.barplot(x=balance_rate.index, y=balance_rate.values)
plt.title('Work-Life Balance vs Employee Attrition')
plt.xlabel('Work-Life Balance Level')
plt.ylabel('Attrition Rate (%)')
plt.savefig(folder + '/worklife_balance_attrition.png',
            dpi=300, bbox_inches='tight')
plt.show()


# 2. Business Travel
travel = pd.crosstab(
    df_clean['BusinessTravel'],
    df_clean['Attrition']
)

travel_rate = travel['Yes'] / travel.sum(axis=1) * 100

plt.figure(figsize=(9, 5))
sns.barplot(x=travel_rate.index, y=travel_rate.values)
plt.title('Business Travel vs Employee Attrition')
plt.xlabel('Business Travel')
plt.ylabel('Attrition Rate (%)')
plt.xticks(rotation=15)
plt.savefig(folder + '/business_travel_attrition.png',
            dpi=300, bbox_inches='tight')
plt.show()


# 3. Age Group
age_groups = pd.cut(
    df_clean['Age'],
    bins=[17, 25, 35, 45, 55, 65],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65']
)

age = pd.crosstab(age_groups, df_clean['Attrition'])

age_rate = age['Yes'] / age.sum(axis=1) * 100

plt.figure(figsize=(9, 5))
sns.barplot(x=age_rate.index, y=age_rate.values)
plt.title('Age Group vs Employee Attrition')
plt.xlabel('Age Group')
plt.ylabel('Attrition Rate (%)')
plt.savefig(folder + '/age_group_attrition.png',
            dpi=300, bbox_inches='tight')
plt.show()


# 4. Employee Tenure
tenure_groups = pd.cut(
    df_clean['YearsAtCompany'],
    bins=[-1, 1, 3, 5, 10, 20, 100],
    labels=['0-1', '2-3', '4-5', '6-10', '11-20', '21+']
)

tenure = pd.crosstab(tenure_groups, df_clean['Attrition'])

tenure_rate = tenure['Yes'] / tenure.sum(axis=1) * 100

plt.figure(figsize=(9, 5))
sns.barplot(x=tenure_rate.index, y=tenure_rate.values)
plt.title('Employee Tenure vs Employee Attrition')
plt.xlabel('Years at Company')
plt.ylabel('Attrition Rate (%)')
plt.savefig(folder + '/employee_tenure_attrition.png',
            dpi=300, bbox_inches='tight')
plt.show()

print("All four missing graphs saved successfully!")
```
